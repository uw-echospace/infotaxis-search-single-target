```python
from pathlib import Path

import pandas as pd

import seaborn as sns
```

```python
from data_prep_plot.plot_utils import bar_violinplot_success_beam_move_res
from data_prep_plot.simulation_data_prep import get_success_fname
```

```python
# Set path
path_data = Path("/Volumes/ssd_2tb_1/infotaxis_simu_2026")

path_fig = Path("../figs_paper_2026")
```

```python
cr_all = [5, 10]
br_all = [1, 2]
bmr_all = [2, 3]
pm_all = [0.001, 0.01, 0.02, 0.05]

for cr in cr_all:
    for br in br_all:
        for bmr in bmr_all:

            path_left = path_data / "20260530_infotaxis_match_summary_compiled_pmax095_pmaxDiff1e-5"
            path_right = path_data / "20260608_infotaxis_beam_movement_restriction_summary_compiled_pmax095_pmaxDiff1e-5"
            pmax_th = 0.95
            pmax_diff_th = 1e-5

            df_x_all = []
            df_y_all = []
            for idx, pm in enumerate(pm_all):
                # Load summary and ph details
                df_x = pd.read_csv(
                    path_left / get_success_fname(
                        search_type="infotaxis", cr=cr, br=br, simulation_type="simple",
                        pmax_th=pmax_th, pmax_diff_th=pmax_diff_th,
                        pm_agent=pm
                    ), index_col=0
                )
                df_y = pd.read_csv(
                    path_right / get_success_fname(
                        search_type="infotaxis", cr=cr, br=br, bmr=bmr, simulation_type="beam_move_res",
                        pmax_th=pmax_th, pmax_diff_th=pmax_diff_th,
                        pm_agent=pm,
                    ), index_col=0
                )
                df_x_all.append(df_x)
                df_y_all.append(df_y)

            fig = bar_violinplot_success_beam_move_res(
                cr=cr, br=br, bmr=bmr, pm_all=pm_all,
                df_x_all=df_x_all, df_y_all=df_y_all,
                x_str="Default", y_str="Restrict beam aim movement",
                stats_type="less",
                colors=sns.color_palette("Set2"),
                colors_dark=sns.color_palette("Dark2"),
                color_idx_left=0, color_idx_right=2,
                print_stats=True,
                comparison_type="simple",
            )

            fig.savefig(
                path_fig / f"fig_restrict_beam_movement_cr{cr}_br{br}_bmr{bmr}_pmaxTh0.95_pmaxDiff1e-5.png",
                dpi=300, bbox_inches="tight"
            )
```
