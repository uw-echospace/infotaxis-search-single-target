In this notebook we prepare data to be plotted in the actual figure generation notebook.

The routines used here are from `20251029_repeat_beam_aim_investigate.ipynb`.

```python
from pathlib import Path

import pandas as pd

from infotaxis import one_target
```

```python
# Path to save figs and data for figs
path_fig = Path("/Users/wujung/code_git/infotaxis/figs_paper")
path_data = Path("/Users/wujung/code_git/infotaxis/notebooks_fig")
```

## Set params

```python
cr_all = list(range(5, 11))
br_all = [1, 2]
```

## Compute the number of grids in beam footprint and search space

```python
num_grid_results_all = []
for cr in cr_all:
    for br in br_all:
        print("-------------------------------------")
        search_dict = None
        param_echo = {
            str(1): {
                "pm_const": 0,
                "pfa_const": 0,
            },
            str(2): {
                "pm_const": 0,
                "pfa_const": 0,
            },
        }

        oth = one_target.OneTargetHex(
            search_rule="infotaxis",
            canvas_radius=cr,
            beam_radius=[br],
            target_cube=(-2, 3, -1),
            aim_start_cube=(0, 0, 0),
            param_animal= param_echo,
            param_echo= param_echo,
            search_dict=search_dict
        )

        grid_in_beam = oth.beam_cover_last_cube.shape[0]
        grid_in_space = oth.canvas_cube.shape[0]

        num_grid_results_all.append([cr, br, grid_in_beam, grid_in_space])

        print(f"cr={cr}, br={br}:")
        print(f" - grid_in_beam: {grid_in_beam}")
        print(f" - grid_in_space: {grid_in_space}")
```

```python
df_grid = pd.DataFrame(
    num_grid_results_all, columns=["cr", "br", "num_grids_in_beam", "num_grids_in_space"]
)
df_grid
```

```python
df_grid.to_csv(path_data / "20251221_num_grid.csv")
```
