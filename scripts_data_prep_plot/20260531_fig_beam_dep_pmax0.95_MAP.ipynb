```python
from pathlib import Path

import pandas as pd

import seaborn as sns
```

```python
from data_prep_plot.plot_utils import bar_violinplot_success_beam_dep
from data_prep_plot.simulation_data_prep import get_success_fname
```

```python
# Set path
path_data = Path("/Volumes/ssd_2tb_1/infotaxis_simu_2026")

path_fig = Path("../figs_paper_2026")
```

```python
cr_all = [5, 10]
br_all = [1, 2]
pm_all = [0.001, 0.002, 0.005, 0.01]

beam_sigma_r = {
    1: 0.5,
    2: 1,
}
beam_scale_all = [3, 5]

for cr in cr_all:
    for br in br_all:
        for beam_scale in beam_scale_all:

            print(f"cr={cr}, br={br}, beam_scale={beam_scale}")

            path_left = path_data / "20260530_MAP_match_summary_compiled_pmax095_pmaxDiff1e-5"
            path_right = path_data / "20260530_MAP_beam_dep_summary_compiled_pmax095_pmaxDiff1e-5"
            pmax_th = 0.95
            pmax_diff_th = 1e-5

            df_x_all = []
            df_y_all = []
            for pm in pm_all:
                # Load summary and ph details
                df_x = pd.read_csv(
                    path_left / get_success_fname(
                        search_type="MAP", cr=cr, br=br, simulation_type="simple",
                        pmax_th=pmax_th, pmax_diff_th=pmax_diff_th,
                        pm_agent=pm
                    ), index_col=0
                )
                df_y = pd.read_csv(
                    path_right / get_success_fname(
                        search_type="MAP", cr=cr, br=br, simulation_type="beam_dep",
                        pmax_th=pmax_th, pmax_diff_th=pmax_diff_th,
                        pm_agent=pm, pm_truth=pm,
                        beam_sigma_r=beam_sigma_r[br], beam_scale=beam_scale
                    ), index_col=0
                )
                df_x_all.append(df_x)
                df_y_all.append(df_y)


            fig = bar_violinplot_success_beam_dep(
                cr=cr, br=br, pm_all=pm_all,
                df_x_all=df_x_all, df_y_all=df_y_all,
                x_str="Constant", y_str="Beam-dependent",
                stats_type="less",
                title_str_additional=f"Beam scaling={beam_scale}",
                colors=sns.color_palette("Set2"),
                colors_dark=sns.color_palette("Dark2"),
                color_idx_left=1, color_idx_right=3,
                print_stats=True,
                crowded_bars=True
            )

            fig.savefig(
                path_fig / f"fig_beam_dep_cr{cr}_br{br}_scale{beam_scale}_pmaxTh0.95_pmaxDiff1e-5_MAP.png",
                dpi=300, bbox_inches="tight"
            )
```
