```python
from pathlib import Path

import numpy as np
import pandas as pd
import xarray as xr
```

```python
path_main = Path("../simulations")

path_data = path_main / "20251113"

path_save = path_main / "20251113_summary"
if not path_save.exists():
    path_save.mkdir(parents=True, exist_ok=True)
```

```python
def load_summary(search_type, cr, br, pm, pfa):
    dates = ["20251113", "20251114"]
    files = []
    for d in dates:
        folder = f"{d}_{search_type}_pmaxRepeat_canvasRadius{cr}_beamRadius{br}_pm{pm:.0e}_pfa{pfa:.0e}"
        files += sorted(list((path_data / folder).glob("*_summary.csv")))

    data_list = []
    for f in files:
        data_list.append(pd.read_csv(f, index_col=0))
    df = pd.concat(data_list).reset_index(drop=True)
    return df
```

```python
def load_ph(search_type, cr, br, pm, pfa):
    dates = ["20251113", "20251114"]
    files = []
    for d in dates:
        folder = f"{d}_{search_type}_pmaxRepeat_canvasRadius{cr}_beamRadius{br}_pm{pm:.0e}_pfa{pfa:.0e}"
        files += sorted(list((path_data / folder).glob("*_ph.csv")))

    p_list = []
    h_list = []
    for f in files:
        df = pd.read_csv(f, index_col=0)
        p_list.append(df["p_max_all"])
        h_list.append(df["h_actual_all"])

    # Pad ragged list to array
    len_max = max(len(h) for h in h_list)
    h_actual_pad = np.array([h.tolist() + [np.nan] * (len_max - len(h)) for h in h_list])
    p_max_pad = np.array([p.tolist() + [np.nan] * (len_max - len(p)) for p in p_list])

    ds = xr.Dataset(
        data_vars=dict(
            h_actual=(["run", "ping"], h_actual_pad),
            p_max=(["run", "ping"], p_max_pad),
        ),
        coords=dict(
            run=("run", range(1, 1001)),
            ping=("ping", range(len_max)),
        ),
    )
    return ds 
```

## PFA sweep

```python
cr_all = [5, 8, 10]
br_all = [1, 2]
pm_all = [0]
pfa_all = [0.001, 0.002, 0.005, 0.01, 0.02, 0.05]
```

```python
# infotaxis ph data
search_type = "infotaxis"
for cr in cr_all:
    for br in br_all:
        for pm in pm_all:
            for pfa in pfa_all:
                ds = load_ph(search_type, cr=cr, br=br, pm=pm, pfa=pfa)
                ds.to_netcdf(path_save / f"{search_type}_cr{cr}_br{br}_pm{pm:.0e}_pfa{pfa:.0e}_ph.nc")
```

```python
# MAP ph data
search_type = "MAP"
for cr in cr_all:
    for br in br_all:
        for pm in pm_all:
            for pfa in pfa_all:
                df = load_ph(search_type, cr=cr, br=br, pm=pm, pfa=pfa)
                df.to_netcdf(path_save / f"{search_type}_cr{cr}_br{br}_pm{pm:.0e}_pfa{pfa:.0e}_ph.nc")
```

```python
# infotaxis summary
search_type = "infotaxis"
for cr in cr_all:
    for br in br_all:
        for pm in pm_all:
            for pfa in pfa_all:
                df = load_summary(search_type, cr=cr, br=br, pm=pm, pfa=pfa)
                df.to_csv(path_save / f"{search_type}_cr{cr}_br{br}_pm{pm:.0e}_pfa{pfa:.0e}_summary.csv")
```

```python
# MAP summary
search_type = "MAP"
for cr in cr_all:
    for br in br_all:
        for pm in pm_all:
            for pfa in pfa_all:
                df = load_summary(search_type, cr=cr, br=br, pm=pm, pfa=pfa)
                df.to_csv(path_save / f"{search_type}_cr{cr}_br{br}_pm{pm:.0e}_pfa{pfa:.0e}_summary.csv")
```
